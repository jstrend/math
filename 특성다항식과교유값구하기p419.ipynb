```python
import sympy as sp


# --------------------------------------------------
# 1. 정사각행렬 입력
# --------------------------------------------------

def input_square_matrix():
    n = int(input("정사각행렬의 차수를 입력하세요: "))

    print(f"\n{n}×{n} 행렬의 각 행을 입력하세요.")
    print("각 원소는 공백으로 구분합니다.")
    print("정수, 소수, 분수(예: 1/2)를 입력할 수 있습니다.\n")

    matrix_data = []

    for i in range(n):
        while True:
            values = input(f"{i + 1}행: ").split()

            if len(values) != n:
                print(f"행렬 원소를 정확히 {n}개 입력하세요.")
                continue

            try:
                row = [sp.Rational(value) for value in values]
                matrix_data.append(row)
                break

            except ValueError:
                print("올바른 숫자 형식으로 다시 입력하세요.")

    return sp.Matrix(matrix_data)


# --------------------------------------------------
# 2. 특성다항식과 고유값 계산
# --------------------------------------------------

def calculate_eigenvalues(A):
    if A.rows != A.cols:
        raise ValueError("특성다항식과 고유값은 정사각행렬에서만 구할 수 있습니다.")

    n = A.rows
    lam = sp.symbols("lambda")

    # 특성행렬: λI - A
    characteristic_matrix = lam * sp.eye(n) - A

    # 특성다항식: det(λI - A)
    characteristic_polynomial = sp.expand(
        characteristic_matrix.det()
    )

    # 특성다항식 인수분해
    factored_polynomial = sp.factor(
        characteristic_polynomial
    )

    # 고유값과 대수적 중복도
    eigenvalues = A.eigenvals()

    # 고유값과 고유벡터
    eigenvectors = A.eigenvects()

    return (
        lam,
        characteristic_matrix,
        characteristic_polynomial,
        factored_polynomial,
        eigenvalues,
        eigenvectors
    )


# --------------------------------------------------
# 3. 프로그램 실행
# --------------------------------------------------

A = input_square_matrix()

(
    lam,
    characteristic_matrix,
    characteristic_polynomial,
    factored_polynomial,
    eigenvalues,
    eigenvectors
) = calculate_eigenvalues(A)


print("\n" + "=" * 55)
print("특성다항식과 고유값 계산 결과")
print("=" * 55)

print("\n① 입력 행렬 A")
sp.pprint(A)

print("\n② 특성행렬 λI - A")
sp.pprint(characteristic_matrix)

print("\n③ 특성다항식")
print("p(λ) = det(λI - A)")

sp.pprint(
    sp.Eq(
        sp.Function("p")(lam),
        characteristic_polynomial
    )
)

print("\n④ 특성다항식의 인수분해")

sp.pprint(
    sp.Eq(
        sp.Function("p")(lam),
        factored_polynomial
    )
)

print("\n⑤ 고유방정식")
sp.pprint(sp.Eq(characteristic_polynomial, 0))

print("\n⑥ 고유값과 대수적 중복도")

for eigenvalue, multiplicity in eigenvalues.items():
    print("\n고유값:")
    sp.pprint(eigenvalue)

    print(f"대수적 중복도: {multiplicity}")

    print("근삿값:")
    sp.pprint(sp.N(eigenvalue, 6))


print("\n⑦ 고유값에 대응하는 고유벡터")

for eigenvalue, multiplicity, vectors in eigenvectors:
    print("\n" + "-" * 40)

    print("고유값:")
    sp.pprint(eigenvalue)

    print(f"대수적 중복도: {multiplicity}")

    print("대응하는 고유벡터의 기저:")

    for vector in vectors:
        sp.pprint(vector)

        # Av = λv 검산
        left_side = sp.simplify(A * vector)
        right_side = sp.simplify(eigenvalue * vector)

        print("검산 결과 Av:")
        sp.pprint(left_side)

        print("검산 결과 λv:")
        sp.pprint(right_side)

        if left_side == right_side:
            print("Av = λv이므로 고유벡터가 정확합니다.")
```
