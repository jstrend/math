```python
import sympy as sp


def input_square_matrix():
    """사용자로부터 정사각행렬을 입력받는 함수"""

    n = int(input("정사각행렬의 차수를 입력하세요: "))
    matrix_data = []

    print(f"\n{n}×{n} 행렬의 각 행을 공백으로 구분하여 입력하세요.")
    print("정수, 소수, 분수(예: 1/2)를 입력할 수 있습니다.\n")

    for i in range(n):
        while True:
            values = input(f"{i + 1}행: ").split()

            if len(values) != n:
                print(f"숫자를 정확히 {n}개 입력하세요.")
                continue

            try:
                row = [sp.Rational(value) for value in values]
                matrix_data.append(row)
                break

            except ValueError:
                print("올바른 숫자 형식으로 다시 입력하세요.")

    return sp.Matrix(matrix_data)


def inverse_using_adjugate(A):
    """
    수반행렬을 이용하여 역행렬 계산

    A^(-1) = adj(A) / det(A)
    """

    # 정사각행렬 확인
    if A.rows != A.cols:
        raise ValueError("정사각행렬만 역행렬을 구할 수 있습니다.")

    # 행렬식 계산
    det_A = A.det()

    # 행렬식이 0이면 역행렬이 존재하지 않음
    if det_A == 0:
        return det_A, None, None, None

    # 여인수 행렬 계산
    cofactor_matrix = A.cofactor_matrix()

    # 여인수 행렬을 전치하여 수반행렬 계산
    adj_A = cofactor_matrix.T

    # 역행렬 계산
    inverse_A = sp.simplify((1 / det_A) * adj_A)

    return det_A, cofactor_matrix, adj_A, inverse_A


# --------------------------------------------------
# 프로그램 실행
# --------------------------------------------------

A = input_square_matrix()

det_A, cofactor_A, adj_A, inverse_A = inverse_using_adjugate(A)

print("\n① 입력 행렬 A")
sp.pprint(A)

print("\n② 행렬식 det(A)")
sp.pprint(det_A)

if det_A == 0:
    print("\n행렬식이 0이므로 역행렬이 존재하지 않습니다.")

else:
    print("\n③ 여인수 행렬 C")
    sp.pprint(cofactor_A)

    print("\n④ 수반행렬 adj(A) = Cᵀ")
    sp.pprint(adj_A)

    print("\n⑤ 역행렬 계산 공식")
    print("A⁻¹ = (1 / det(A)) × adj(A)")

    print("\n⑥ 역행렬 A⁻¹")
    sp.pprint(inverse_A)

    # 역행렬 검산
    verification = sp.simplify(A * inverse_A)

    print("\n⑦ 검산: A × A⁻¹")
    sp.pprint(verification)

    if verification == sp.eye(A.rows):
        print("\n검산 결과: 단위행렬이므로 역행렬이 정확합니다.")
    else:
        print("\n검산 결과: 계산 결과를 확인해야 합니다.")
```
