```python
# 여인수 전개를 이용한 행렬식 계산

def minor_matrix(matrix, row, col):
    """지정된 행과 열을 제거한 소행렬을 반환"""
    return [
        [matrix[i][j] for j in range(len(matrix)) if j != col]
        for i in range(len(matrix)) if i != row
    ]


def determinant(matrix):
    """여인수 전개를 이용하여 행렬식을 재귀적으로 계산"""
    n = len(matrix)

    # 정사각행렬 확인
    if n == 0 or any(len(row) != n for row in matrix):
        raise ValueError("정사각행렬만 입력할 수 있습니다.")

    # 1×1 행렬
    if n == 1:
        return matrix[0][0]

    # 2×2 행렬
    if n == 2:
        return (
            matrix[0][0] * matrix[1][1]
            - matrix[0][1] * matrix[1][0]
        )

    # 첫 번째 행을 기준으로 여인수 전개
    det = 0

    for col in range(n):
        minor = minor_matrix(matrix, 0, col)
        cofactor_sign = (-1) ** col
        cofactor = cofactor_sign * determinant(minor)

        det += matrix[0][col] * cofactor

    return det


# 행렬 입력
n = int(input("정사각행렬의 차수를 입력하세요: "))

matrix = []

print(f"{n}×{n} 행렬의 각 행을 공백으로 구분하여 입력하세요.")

for i in range(n):
    while True:
        row = list(map(float, input(f"{i + 1}행: ").split()))

        if len(row) == n:
            matrix.append(row)
            break

        print(f"숫자를 {n}개 입력해야 합니다.")


# 결과 출력
result = determinant(matrix)

print("\n입력 행렬")
for row in matrix:
    print(row)

print("\n여인수 전개로 구한 행렬식 =", result)
```
